```python
import numpy as np
from scipy.sparse import diags
import time
import matplotlib.pyplot as plt
```

```python
f = 1
psi = 1
x = np.linspace(0.0001,0.9999999,1000)
epsilons = [1,0.01,0.001]
fig, ax = plt.subplots()
for eps in epsilons:
    ys = 1/psi*((1+(np.exp(psi/eps)-1)*(x)-np.exp(x*(psi/eps)))/(np.exp(psi/eps)-1))
    ax.plot(x,ys,label = f"Epsilon= {eps}")
ax.legend()
plt.show()
```
